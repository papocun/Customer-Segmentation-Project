1. Missing Data
Income has 24 missing values: Its count is 2216, while all other columns have 2240. We need to handle these missing values before analyzing or modeling.

2. Major Outliers & Anomalies
Income has a massive outlier: The maximum income is 666,666, while 75% of your data lies below 68,522. This single extreme value is distorting your mean.

Year_Birth has impossible data: The minimum birth year is 1893. Unless a customer is over 130 years old, this is a typo or placeholder data that needs to be removed.

3. Heavily Skewed Columns
Spending data (Mnt...) is heavily right-skewed: For columns like MntWines and MntMeatProducts, the mean is much higher than the median (50%), and the maximum values are huge.

Customer behavior: This shows that most customers spend small amounts, while a tiny group of "high rollers" spend a lot.

4. Discrete Columns
Kidhome and Teenhome are discrete counts: They range strictly from 0 to 2. Even though they are shown as decimals, they represent distinct household groups (0, 1, or 2 kids).

```python

#missing values information 
missing = df.isnull().sum()
missing_pct = (missing / len(df)) * 100
missing_df = pd.DataFrame({
    'Missing Count': missing,
    'Missing %': missing_pct.round(2)
}).query('`Missing Count` > 0')

print('Columns with missing values:')
print(missing_df if len(missing_df) > 0 else 'None — dataset is complete!')
```

## Step 2 - Data Cleaning 

```python
#Fixing the date column

df['Dt_Customer'] = pd.to_datetime(df['Dt_Customer'], format='%d-%m-%Y')
print('Dt_Customer dtype:', df['Dt_Customer'].dtype)
print('Date range:', df['Dt_Customer'].min().date(), 'to', df['Dt_Customer'].max().date())
```

```python
df.dtypes
```

```python
# Define numeric and categorical column lists
columns = [col for col in df.columns if col not in ['ID', 'Dt_Customer']]

numeric_features    = [f for f in columns if df[f].dtype != 'O']
categorical_features = [f for f in columns if df[f].dtype == 'O']

print(f'Numeric features ({len(numeric_features)}): {numeric_features}')
print(f'\nCategorical features ({len(categorical_features)}): {categorical_features}')
```

```python
df.nunique()
```

```python
# Drop zero-variance columns
# Z_CostContact is always 3, Z_Revenue is always 11 — confirmed with .nunique()
print('Z_CostContact unique values:', df['Z_CostContact'].unique())
print('Z_Revenue unique values    :', df['Z_Revenue'].unique())

df.drop(columns=['Z_CostContact', 'Z_Revenue'], inplace=True)

# also drop ID — unique per row, no analytical value
df.drop(columns=['ID'], inplace=True)

print('\nDropped: Z_CostContact, Z_Revenue, ID')
print('Remaining columns:', df.shape[1])
```

```python
#As income is right skewed we are preferring to use median instead of mean for missing values imputation
median_income = df['Income'].median()
print(f'Median income: ${median_income:,.0f}')
print(f'Missing income rows before fill: {df["Income"].isnull().sum()}')

df['Income'] = df['Income'].fillna(median_income)

print(f'Missing income rows after fill:  {df["Income"].isnull().sum()}')
```

```python
#Clean Marital_Status: consolidate rare/unusual categories
print('Before:', df['Marital_Status'].value_counts().to_dict())

# Alone, Absurd, YOLO are clearly data quality issues — group as 'Single'
df['Marital_Status'] = df['Marital_Status'].replace({
    'Alone': 'Single',
    'Absurd': 'Single',
    'YOLO': 'Single'
})

print('After:', df['Marital_Status'].value_counts().to_dict())
```

```python
#Remove extreme outliers in Age and Income
# Anyone born before 1933 (age > 89) is almost certainly a data entry error
# Income of 666,666 is clearly erroneous (10x the next highest value)

print('Age outliers (Year_Birth < 1933):',  (df['Year_Birth'] < 1933).sum())
print('Income outlier (Income > 600000):', (df['Income'] > 600000).sum())

df = df[df['Year_Birth'] >= 1933]
df = df[df['Income'] < 600000]

print(f'\nClean dataset shape: {df.shape}')
```

```python
#Check duplicates
dupes = df.duplicated().sum()
print(f'Duplicate rows: {dupes}')
if dupes > 0:
    df.drop_duplicates(inplace=True)
    print('Duplicates removed.')
```

### Feature Information

- **People**

    - **ID:** Customer's unique identifier
    - **Year_Birth:** Customer's birth year
    - **Education:** Customer's education level
    - **Marital_Status:** Customer's marital status
    - **Income:** Customer's yearly household income
    - **Kidhome:** Number of children in customer's household
    - **Teenhome:** Number of teenagers in customer's household
    - **Dt_Customer:** Date of customer's enrollment with the company
    - **Recency:** Number of days since customer's last purchase
    - **Complain:** 1 if the customer complained in the last 2 years, 0 otherwise

- **Products**

    - **MntWines:** Amount spent on wine in last 2 years
    - **MntFruits:** Amount spent on fruits in last 2 years
    - **MntMeatProducts:** Amount spent on meat in last 2 years
    - **MntFishProducts:** Amount spent on fish in last 2 years
    - **MntSweetProducts:** Amount spent on sweets in last 2 years
    - **MntGoldProds:** Amount spent on gold in last 2 years

- **Promotion**

    - **NumDealsPurchases:** Number of purchases made with a discount
    - **AcceptedCmp1:** 1 if customer accepted the offer in the 1st campaign, 0 otherwise
    - **AcceptedCmp2:** 1 if customer accepted the offer in the 2nd campaign, 0 otherwise
    - **AcceptedCmp3:** 1 if customer accepted the offer in the 3rd campaign, 0 otherwise
    - **AcceptedCmp4:** 1 if customer accepted the offer in the 4th campaign, 0 otherwise
    - **AcceptedCmp5:** 1 if customer accepted the offer in the 5th campaign, 0 otherwise
    - **Response:** 1 if customer accepted the offer in the last campaign, 0 otherwise

- **Place**

    - **NumWebPurchases:** Number of purchases made through the company’s website
    - **NumCatalogPurchases:** Number of purchases made using a catalogue
    - **NumStorePurchases:** Number of purchases made directly in stores
    - **NumWebVisitsMonth:** Number of visits to company’s website in the last month

## Step 3 - Feature Engineering 

```python
#Age
df['Age'] = 2022 - df['Year_Birth']
print('Age range:', df['Age'].min(), 'to', df['Age'].max())
print('Mean age :', round(df['Age'].mean(), 1))
```

```python
# Children in household
df['Children'] = df['Kidhome'] + df['Teenhome']

# Has a partner (binary: 1=with partner, 0=alone)
df['Has_Partner'] = df['Marital_Status'].isin(['Married', 'Together']).astype(int)

# Family size
df['Family_Size'] = df['Has_Partner'] + df['Children'] + 1

#  Is a parent?
df['Is_Parent'] = (df['Children'] > 0).astype(int)

print('Children distribution:')
print(df['Children'].value_counts().sort_index())
print('\nFamily size distribution:')
print(df['Family_Size'].value_counts().sort_index())
```

```python
# Spending features
spend_cols = ['MntWines','MntFruits','MntMeatProducts',
              'MntFishProducts','MntSweetProducts','MntGoldProds']

df['Total_Spending'] = df[spend_cols].sum(axis=1)

# Spend ratio = what fraction of income is spent
df['Spend_Ratio'] = (df['Total_Spending'] / df['Income']).round(4)

print('Total spending stats:')
print(df['Total_Spending'].describe())
```

```python
# Promotion responsiveness
promo_cols = ['AcceptedCmp1','AcceptedCmp2','AcceptedCmp3','AcceptedCmp4','AcceptedCmp5']
df['Total_Promo_Accepted'] = df[promo_cols].sum(axis=1)
df['Any_Promo_Accepted']   = (df['Total_Promo_Accepted'] > 0).astype(int)

print('Customers who accepted at least 1 promo:', df['Any_Promo_Accepted'].sum())
print('Overall promo acceptance rate: {:.1f}%'.format(
    df['Any_Promo_Accepted'].mean() * 100))
```

```python
#  Total purchases across all channels
purchase_cols = ['NumWebPurchases','NumCatalogPurchases','NumStorePurchases','NumDealsPurchases']
df['Total_Purchases'] = df[purchase_cols].sum(axis=1)

# Days as customer (from enrolment date to 2022-01-01)
reference_date = pd.Timestamp('2022-01-01')
df['Days_as_Customer'] = (reference_date - df['Dt_Customer']).dt.days
df['Year_Joined'] = df['Dt_Customer'].dt.year

# Age buckets
df['Age_Group'] = pd.cut(df['Age'],
    bins=[0, 35, 50, 65, 100],
    labels=['Young (<35)', 'Middle (35-50)', 'Senior (50-65)', 'Elderly (65+)'])

#  Income bucket
df['Income_Group'] = pd.cut(df['Income'],
    bins=[0, 30000, 60000, 90000, 200000],
    labels=['Low (<30k)', 'Mid (30-60k)', 'High (60-90k)', 'Very High (90k+)'])

print('New features added. Final shape:', df.shape)
```

```python
# Rename spending columns for readability
df.rename(columns={
    'MntWines': 'Wines', 'MntFruits': 'Fruits',
    'MntMeatProducts': 'Meat', 'MntFishProducts': 'Fish',
    'MntSweetProducts': 'Sweets', 'MntGoldProds': 'Gold',
    'NumWebPurchases': 'Web_Purchases',
    'NumCatalogPurchases': 'Catalog_Purchases',
    'NumStorePurchases': 'Store_Purchases',
    'NumDealsPurchases': 'Deal_Purchases'
}, inplace=True)

# drop original columns now replaced by derived ones
df.drop(columns=['Year_Birth', 'Kidhome', 'Teenhome'], inplace=True)

print('Final columns:', df.columns.tolist())
```

## Step 4 - Univariate Analysis

```python
# Age distribution
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle('Customer Age Distribution', fontsize=15, fontweight='bold', y=1.02)

# histogram + KDE
sns.histplot(df['Age'], bins=40, kde=True, color='#2E86AB', ax=axes[0])
axes[0].axvline(df['Age'].mean(), color='red', linestyle='--', label=f"Mean: {df['Age'].mean():.1f}")
axes[0].axvline(df['Age'].median(), color='orange', linestyle='--', label=f"Median: {df['Age'].median():.1f}")
axes[0].set_title('Age Histogram with KDE')
axes[0].set_xlabel('Age')
axes[0].legend()

# boxplot
sns.boxplot(y=df['Age'], color='#2E86AB', ax=axes[1])
axes[1].set_title('Age Boxplot')
axes[1].set_ylabel('Age')

plt.tight_layout()
plt.show()

print('Age Stats:')
print(df['Age'].describe().round(1))
```

```python
#Income distribution
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle('Customer Income Distribution', fontsize=15, fontweight='bold')

sns.histplot(df['Income'], bins=50, kde=True, color='#44BBA4', ax=axes[0])
axes[0].axvline(df['Income'].median(), color='red', linestyle='--',
                label=f"Median: ${df['Income'].median():,.0f}")
axes[0].set_title('Income Histogram (right-skewed)')
axes[0].set_xlabel('Income ($)')
axes[0].legend()

sns.boxplot(y=df['Income'], color='#44BBA4', ax=axes[1])
axes[1].set_title('Income Boxplot')

plt.tight_layout()
plt.show()

print(f'Skewness: {df["Income"].skew():.2f}  (positive = right-skewed)')
```

```python
#Total spending distribution
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle('Total Spending Distribution', fontsize=15, fontweight='bold')

sns.histplot(df['Total_Spending'], bins=50, kde=True, color='#A23B72', ax=axes[0])
axes[0].axvline(df['Total_Spending'].median(), color='red', linestyle='--',
                label=f"Median: ${df['Total_Spending'].median():,.0f}")
axes[0].set_title('Total Spending Histogram')
axes[0].legend()

sns.boxplot(y=df['Total_Spending'], color='#A23B72', ax=axes[1])
axes[1].set_title('Total Spending Boxplot')

plt.tight_layout()
plt.show()
```

```python
#Spending by category (pie chart)
spend_cols = ['Wines', 'Fruits', 'Meat', 'Fish', 'Sweets', 'Gold']
spend_totals = df[spend_cols].sum()

colors = ['#2E86AB','#44BBA4','#A23B72','#F18F01','#C73E1D','#7B2D8B']

fig, axes = plt.subplots(1, 2, figsize=(14, 6))
fig.suptitle('Spending by Product Category', fontsize=15, fontweight='bold')

wedges, texts, autotexts = axes[0].pie(
    spend_totals,
    labels=spend_cols,
    autopct='%1.1f%%',
    colors=colors,
    startangle=140,
    wedgeprops={'edgecolor': 'white', 'linewidth': 2}
)
for t in autotexts:
    t.set_fontsize(10)
axes[0].set_title('Share of Total Spend')

sns.barplot(x=spend_cols, y=spend_totals.values, palette=colors, ax=axes[1])
axes[1].set_title('Total Amount Spent per Category ($)')
axes[1].set_xlabel('Category')
axes[1].set_ylabel('Total Spend ($)')
for p in axes[1].patches:
    axes[1].annotate(f'${p.get_height():,.0f}',
                     (p.get_x() + p.get_width()/2, p.get_height() + 5000),
                     ha='center', fontsize=9)

plt.tight_layout()
plt.show()
```
